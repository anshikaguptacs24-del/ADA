## Tic Tac Toe Game (Two Players)

```python
def display_board(board):
    """Prints the Tic Tac Toe board."""
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("-----------")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("-----------")
    print(f" {board[6]} | {board[7]} | {board[8]} ")

def player_input(player_marker, board):
    """Gets valid player input for their move."""
    position = -1
    while position not in range(1, 10) or not is_space_free(board, position - 1):
        try:
            position = int(input(f"Player {player_marker}, choose your next position (1-9): "))
        except ValueError:
            print("Invalid input. Please enter a number.")
    return position - 1 # Adjust for 0-based indexing

def place_marker(board, marker, position):
    """Places the player's marker on the board."""
    board[position] = marker

def is_space_free(board, position):
    """Checks if a position on the board is free."""
    return board[position] == ' '

def check_win(board, marker):
    """Checks if the current player has won."""
    win_conditions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8), # Rows
        (0, 3, 6), (1, 4, 7), (2, 5, 8), # Columns
        (0, 4, 8), (2, 4, 6)             # Diagonals
    ]
    for condition in win_conditions:
        if all(board[i] == marker for i in condition):
            return True
    return False

def is_board_full(board):
    """Checks if the board is full (for a draw)."""
    return ' ' not in board

def play_game():
    """Main function to run the Tic Tac Toe game."""
    board = [' '] * 9
    player1_marker = 'X'
    player2_marker = 'O'
    current_player = player1_marker
    game_on = True

    print("Welcome to Tic Tac Toe!")
    display_board(board)

    while game_on:
        position = player_input(current_player, board)
        place_marker(board, current_player, position)
        display_board(board)

        if check_win(board, current_player):
            print(f"Congratulations, Player {current_player} has won!")
            game_on = False
        elif is_board_full(board):
            print("The game is a draw!")
            game_on = False
        else:
            # Switch players
            current_player = player2_marker if current_player == player1_marker else player1_marker

    print("Game Over!")

# To start the game, uncomment and run the line below:
play_game()
```

## Vacuum Cleaner Simulation

```python
import time

class Room:
    def __init__(self, name, is_dirty):
        self.name = name
        self.is_dirty = is_dirty

    def __str__(self):
        status = "Dirty" if self.is_dirty else "Clean"
        return f"Room {self.name}: {status}"

    def clean(self):
        if self.is_dirty:
            print(f"  Cleaning Room {self.name}...")
            time.sleep(1) # Simulate cleaning time
            self.is_dirty = False
            print(f"  Room {self.name} is now Clean.")
        else:
            print(f"  Room {self.name} is already Clean.")

class VacuumCleaner:
    def __init__(self, rooms):
        self.rooms = rooms
        self.current_room_index = 0

    def operate(self):
        print("Starting vacuum cleaner operation...")
        for room in self.rooms:
            print(f"\nVacuum cleaner is in {room.name}.")
            if room.is_dirty:
                room.clean()
            else:
                print(f"  Room {room.name} is clean, moving to the next room.")
            time.sleep(0.5)
        print("\nVacuum cleaner operation finished.")
```

```python
# Initialize two rooms with different statuses
room1 = Room("A", is_dirty=True)
room2 = Room("B", is_dirty=True)

# Display initial room states
print("Initial Room States:")
print(room1)
print(room2)

# Create a vacuum cleaner and run the simulation
vacuum = VacuumCleaner([room1, room2])
vacuum.operate()

# Display final room states
print("\nFinal Room States:")
print(room1)
print(room2)
```
